```python
def calculate_annual_tax(annual_salary):

    # Annual income tax slabs
    slabs = [
        (0, 600000, 0.00),
        (600000, 1200000, 0.01),
        (1200000, 2200000, 0.11),
        (2200000, 3200000, 0.20),
        (3200000, 4100000, 0.25),
        (4100000, 5600000, 0.29),
        (5600000, 7000000, 0.32),
        (7000000, None, 0.35)
    ]

    total_tax = 0
    slab_details = []

    for lower, upper, rate in slabs:

        if annual_salary <= lower:
            break

        if upper is None:
            taxable_amount = annual_salary - lower
            slab_name = f"Above Rs. {lower:,.0f}"

        else:
            taxable_amount = min(annual_salary, upper) - lower
            slab_name = (
                f"Rs. {lower:,.0f} - Rs. {upper:,.0f}"
            )

        slab_tax = taxable_amount * rate
        total_tax += slab_tax

        slab_details.append({
            "slab": slab_name,
            "amount": taxable_amount,
            "rate": rate,
            "tax": slab_tax
        })

    return total_tax, slab_details


def find_gross_salary(net_annual_salary):

    low = net_annual_salary
    high = net_annual_salary * 2 + 1000000

    for _ in range(100):

        gross_salary = (low + high) / 2

        tax, _ = calculate_annual_tax(gross_salary)

        salary_after_tax = gross_salary - tax

        if salary_after_tax < net_annual_salary:
            low = gross_salary
        else:
            high = gross_salary

    return (low + high) / 2


def salary_tax_calculator():

    print("=" * 65)
    print("                 SALARY TAX CALCULATOR")
    print("                    Pakistan 2026-27")
    print("=" * 65)

    print("\nSelect the type of salary amount you have:\n")

    print("1. Annual Salary (Before Tax)")
    print("2. Monthly Salary (After Tax)")
    print("3. Monthly Salary (Before Tax)")

    try:

        choice = input("\nEnter your choice (1, 2, or 3): ")

        if choice not in ["1", "2", "3"]:
            print("\nInvalid choice! Please select 1, 2, or 3.")
            return

        salary = float(input("Enter the salary amount (PKR): "))

        if salary < 0:
            print("\nError: Salary cannot be negative.")
            return

        # Option 1: Annual salary before tax
        if choice == "1":

            annual_salary = salary

        # Option 2: Monthly salary after tax
        elif choice == "2":

            net_annual_salary = salary * 12

            annual_salary = find_gross_salary(
                net_annual_salary
            )

        # Option 3: Monthly salary before tax
        else:

            annual_salary = salary * 12

        # Calculate annual tax and slab details
        annual_tax, slab_details = calculate_annual_tax(
            annual_salary
        )

        # Calculate monthly figures
        monthly_salary_before_tax = annual_salary / 12
        monthly_tax = annual_tax / 12

        monthly_salary_after_tax = (
            monthly_salary_before_tax - monthly_tax
        )

        annual_salary_after_tax = annual_salary - annual_tax

        # Calculate effective tax rate
        if annual_salary > 0:
            tax_percentage = (
                annual_tax / annual_salary
            ) * 100
        else:
            tax_percentage = 0

        # Display salary details
        print("\n" + "=" * 65)
        print("                       SALARY DETAILS")
        print("=" * 65)

        if choice == "1":
            print("Input Type: Annual Salary (Before Tax)")

        elif choice == "2":
            print("Input Type: Monthly Salary (After Tax)")

        else:
            print("Input Type: Monthly Salary (Before Tax)")

        print("-" * 65)

        print(
            f"Monthly Salary Before Tax : "
            f"Rs. {monthly_salary_before_tax:,.2f}"
        )

        print(
            f"Monthly Tax Deduction     : "
            f"Rs. {monthly_tax:,.2f}"
        )

        print(
            f"Monthly Salary After Tax  : "
            f"Rs. {monthly_salary_after_tax:,.2f}"
        )

        print("-" * 65)

        print(
            f"Annual Salary Before Tax  : "
            f"Rs. {annual_salary:,.2f}"
        )

        print(
            f"Annual Income Tax         : "
            f"Rs. {annual_tax:,.2f}"
        )

        print(
            f"Annual Salary After Tax   : "
            f"Rs. {annual_salary_after_tax:,.2f}"
        )

        print(
            f"Effective Tax Rate        : "
            f"{tax_percentage:.2f}%"
        )

        # Display detailed tax slab breakdown
        print("\n" + "=" * 65)
        print("                 TAX SLAB BREAKDOWN")
        print("=" * 65)

        print(
            f"{'Income Slab':<27}"
            f"{'Amount in Slab':>18}"
            f"{'Rate':>9}"
            f"{'Tax':>14}"
        )

        print("-" * 68)

        for detail in slab_details:

            print(
                f"{detail['slab']:<27}"
                f"{detail['amount']:>18,.2f}"
                f"{detail['rate'] * 100:>8.0f}%"
                f"{detail['tax']:>14,.2f}"
            )

        print("-" * 68)

        print(
            f"{'TOTAL TAX':<54}"
            f"{annual_tax:>14,.2f}"
        )

        print("=" * 68)

        print("\nTax calculation completed successfully!")

    except ValueError:

        print("\nInvalid input! Please enter a valid number.")


salary_tax_calculator()
```
